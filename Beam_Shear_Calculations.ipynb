# Beam Shear Calculations with PyCivilCalcs

This notebook demonstrates a simple **simply supported beam shear workflow** using `EngEnv` from this repository.

It covers:
- support reactions for a uniform load
- shear at any section `x`
- maximum shear for the span


```python
from re_lib.eng_var import EngEnv

v = EngEnv(auto_display=False)
```

## 1) Define beam loads and geometry

Given a simply supported beam with a full-span uniform load `w` and length `L`:

- Reactions: `R_A = w*L/2`, `R_B = w*L/2`
- Shear function (from left support): `V(x) = R_A - w*x`

```python
# Inputs
v.w = (1.2, 'kip/ft', 'Uniform factored load')
v.L = (30, 'ft', 'Beam span')

# Location to evaluate shear
v.x = (8, 'ft', 'Distance from left support')

v.w.show()
v.L.show()
v.x.show()
```

## 2) Define and evaluate reaction and shear equations

```python
v.eq.R_A = 'w * L / 2'
v.eq.R_B = 'w * L / 2'
v.eq.V_x = 'w * L / 2 - w * x'

v.eq.R_A.show(out_units='kip')
v.eq.R_B.show(out_units='kip')
v.eq.V_x.show(out_units='kip')
```

## 3) Shear at critical locations

```python
# Left support shear (x = 0)
v.x = (0, 'ft')
V_left = v.eq.V_x.value('kip')

# Midspan shear (x = L/2)
v.x = (15, 'ft')
V_mid = v.eq.V_x.value('kip')

# Right support just to the left (x = L)
v.x = (30, 'ft')
V_right = v.eq.V_x.value('kip')

print(f'V(0)      = {V_left:0.3f} kip')
print(f'V(L/2)    = {V_mid:0.3f} kip')
print(f'V(L-)     = {V_right:0.3f} kip')
print(f'V_max abs = {max(abs(V_left), abs(V_right)):0.3f} kip')
```

## 4) Optional: tabulate shear along the span

This gives a quick shear diagram data table using equally spaced points.

```python
x_vals = [0, 5, 10, 15, 20, 25, 30]
rows = []
for xi in x_vals:
    v.x = (xi, 'ft')
    rows.append((xi, v.eq.V_x.value('kip')))

for xi, Vi in rows:
    print(f'x = {xi:>5.1f} ft  ->  V = {Vi:>7.3f} kip')
```
